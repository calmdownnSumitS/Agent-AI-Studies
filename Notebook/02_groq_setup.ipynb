```python
import os
import json
from dotenv import load_dotenv
from openai import OpenAI
from IPython.display import Markdown, display
```

```python
# Always remember to do this!
load_dotenv("../.env")
```

```python

GEMINI_BASE_URL = "https://generativelanguage.googleapis.com/v1beta/openai/"
GROQ_BASE_URL = "https://api.groq.com/openai/v1"
OPENROUTER_BASE_URL = "https://openrouter.ai/api/v1"
```

```python
groq_api_key = os.getenv('GROQ_API_KEY')
openrouter_api_key = os.getenv('OPENROUTER_API_KEY')
```

```python

#add your own key . by signup from given LLMS consoles
gemini = OpenAI(api_key=os.getenv('OPENROUTER_API_KEY'), base_url=GEMINI_BASE_URL)
groq = OpenAI(api_key=os.getenv('OPENROUTER_API_KEY'), base_url=GROQ_BASE_URL)
openrouter = OpenAI(api_key=os.getenv('OPENROUTER_API_KEY'), base_url=OPENROUTER_BASE_URL)
```

```python
question = "how many years it took to count stars its not funny .I am serious about this question I need a exact time for human being in years"

messages = [{"role": "user", "content": question}]
```

```python
competitors = []
answers = []
```

```python
def returValue(modelname,answer):
    competitors.append(modelname)
    answers.append(answer)
    display(Markdown(answer))
```

```python
#gemini

model_name = "gemini-3.6-flash"
response = gemini.chat.completions.create(model=model_name, messages=messages)
answer = response.choices[0].message.content

returValue(model_name, answer)
```

```python
model_name = "llama-3.1-8b-instant"
response = groq.chat.completions.create(model=model_name, messages=messages)
answer = response.choices[0].message.content

returValue(model_name, answer)
```

```python
model_name = "inclusionai/ling-3.0-tiny:free"
response = openrouter.chat.completions.create(model=model_name, messages=messages)
answer = response.choices[0].message.content

returValue(model_name, answer)
```
